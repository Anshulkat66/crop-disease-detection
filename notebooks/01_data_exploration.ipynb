```python
from pathlib import Path

train_dir = Path("data/train")
valid_dir = Path("data/valid")
test_dir = Path("data/test")

train_classes = sorted([d.name for d in train_dir.iterdir() if d.is_dir()])
valid_classes = sorted([d.name for d in valid_dir.iterdir() if d.is_dir()])

print("Train classes:", len(train_classes))
print("Validation classes:", len(valid_classes))

print("\nFirst 5 classes:")
for class_name in train_classes[:5]:
    print(class_name)

train_images = sum(1 for _ in train_dir.rglob("*") if _.is_file())
valid_images = sum(1 for _ in valid_dir.rglob("*") if _.is_file())
test_images = sum(1 for _ in test_dir.rglob("*") if _.is_file())

print("\nTrain images:", train_images)
print("Validation images:", valid_images)
print("Test images:", test_images)
```
